### Covariance and Correlation (Manual)

Rules:
- Do not use numpy or statistics libraries
- Use loops and basic arithmetic only
- Understand the logic step by step


```python
# Two datasets
X = [2, 4, 6, 8]
Y = [1, 3, 5, 7]

n = len(X)

# Mean of X
sum_x = 0
for x in X:
    sum_x = sum_x + x
mean_x = sum_x / n

# Mean of Y
sum_y = 0
for y in Y:
    sum_y = sum_y + y
mean_y = sum_y / n

mean_x, mean_y
```

```python
# Covariance = sum((xi - mean_x)(yi - mean_y)) / n

cov_sum = 0

for i in range(n):
    cov_sum = cov_sum + (X[i] - mean_x) * (Y[i] - mean_y)

covariance = cov_sum / n
covariance
```

```python
# Standard deviation of X
var_x_sum = 0
for x in X:
    var_x_sum = var_x_sum + (x - mean_x) ** 2
std_x = (var_x_sum / n) ** 0.5

# Standard deviation of Y
var_y_sum = 0
for y in Y:
    var_y_sum = var_y_sum + (y - mean_y) ** 2
std_y = (var_y_sum / n) ** 0.5

std_x, std_y
```

```python
# Correlation = covariance / (std_x * std_y)

correlation = covariance / (std_x * std_y)
correlation
```

### Explanation

- Covariance measures how two variables change together
- Positive covariance means they increase together
- Correlation standardises covariance
- Correlation values range from -1 to 1
- All calculations are done manually using loops and indices
