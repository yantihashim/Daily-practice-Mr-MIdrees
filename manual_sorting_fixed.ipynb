### Explore how to sort a list manually

Rules:
- Do not use list methods like `sort()`
- Do not use built-in sorting libraries
- Understand the logic step by step


```python
# Manual sorting using nested loops
# No list methods are used

numbers = [5, 20, 30, 80, 15]
n = len(numbers)

for i in range(n):
    for j in range(i + 1, n):
        if numbers[i] > numbers[j]:
            temp = numbers[i]
            numbers[i] = numbers[j]
            numbers[j] = temp

numbers
```

### Explanation

- We compare two elements at a time
- If the left element is greater than the right one, we swap them
- We repeat this process until the list is sorted
- This helps us understand how sorting works internally
