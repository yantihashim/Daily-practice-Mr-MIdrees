### Mean, Variance, and Standard Deviation (Manual)

Rules:
- Do not use numpy or statistics libraries
- Use loops and basic arithmetic only
- Understand each step clearly


```python
numbers = [5, 20, 30, 80, 15]

total = 0
n = len(numbers)

for x in numbers:
    total = total + x

mean = total / n
mean
```

```python
# Variance = sum((x - mean)^2) / n

variance_sum = 0

for x in numbers:
    diff = x - mean
    variance_sum = variance_sum + diff * diff

variance = variance_sum / n
variance
```

```python
# Standard deviation = sqrt(variance)

std = variance ** 0.5
std
```

### Explanation

- Mean gives the central value
- Variance measures how spread out the data is
- Standard deviation is the square root of variance
- All calculations are done manually using loops
